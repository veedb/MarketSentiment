```python
import requests
from bs4 import BeautifulSoup

# The URL of a news site (Example: Google News / Finance)
url = "https://news.google.com/topics/CAAqJggKIiBDQkFTRWdvSUwyMHZNRGx6TVdZU0FtVnVHZ0pWVXlnQVAB?hl=en-IN&gl=IN&ceid=IN%3Aen"

# Fetch the page content
response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

# Find all the headlines (Google News uses 'article' tags)
articles = soup.find_all('a', class_='gPFEn') # This class name might change, but it works for now

live_headlines = []
for item in articles[:10]: # Just take the first 10 headlines
    live_headlines.append(item.text)

print(f"Successfully grabbed {len(live_headlines)} live headlines!")
for h in live_headlines:
    print("-", h)
```
